### Problem 20

Compute a singular value decomposition $$\mathbf{A} = \begin{bmatrix} 3 & 1 & 0 & 0 \\ 1 & 3 & 0 & 0 \\ 0 & 0 & 2 & 0 \end{bmatrix} = \mathbf{U} \mathbf{\Sigma} \mathbf{V}^\top.$$
Set `p20u` to $\mathbf{U}$, `p20s` to $\mathbf{\Sigma}$, and `p20v` to $\mathbf{V}$.

Hint: `np.linalg` can help with this, but does not return the standard matrix sizes.

```python
# DO NOT CHANGE

A20 = np.array([[3, 1, 0, 0], [1, 3, 0, 0], [0, 0, 2, 0]])
```

```python
# YOUR CHANGES HERE

p20u, s, v = np.linalg.svd(A20)

# Construct the 3 x 4 matrix to p20s
p20s = np.zeros((3, 4))
np.fill_diagonal(p20s, s)

# Transpose V to p20v
p20v = v.T
```
